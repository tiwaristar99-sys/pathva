```
import torch
from diffusers import LTXPipeline
from diffusers.utils import export_to_video

pipe = LTXPipeline.from_pretrained("Lightricks/LTX-Video", torch_dtype=torch.bfloat16)
pipe.enable_model_cpu_offload()
pipe.vae.enable_tiling()
```

```
# YE APNA PROMPT DAALO
prompt = "A cinematic shot of a busy Indian street at dusk, camera gliding forward, warm streetlights"

video = pipe(
    prompt=prompt,
    negative_prompt="worst quality, blurry, distorted",
    width=704, height=480,          # keep multiples of 32
    num_frames=121,                  # ~5 seconds at 24fps
    num_inference_steps=40,
    guidance_scale=3.0,
).frames[0]

export_to_video(video, "output.mp4", fps=24)
print("done -> output.mp4")
```
